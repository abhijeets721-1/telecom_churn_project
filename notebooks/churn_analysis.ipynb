### Data Cleaning

```python
df = df_customer.copy()
```

```python
# Offer: null means the customer did not accept any offer
df["Offer"] = df["Offer"].fillna("No Offer")

# Customers without phone service
df["Avg Monthly Long Distance Charges"] = df["Avg Monthly Long Distance Charges"].fillna(0)
df["Multiple Lines"] = df["Multiple Lines"].fillna("No")

# Customers without internet service
df["Internet Type"] = df["Internet Type"].fillna("None")
df["Avg Monthly GB Download"] = df["Avg Monthly GB Download"].fillna(0)

internet_addons = ["Online Security", "Online Backup", "Device Protection Plan",
                   "Premium Tech Support", "Streaming TV", "Streaming Movies",
                   "Streaming Music", "Unlimited Data"]
df[internet_addons] = df[internet_addons].fillna("No")

# Churn reason exists only for churned customers
df["Churn Category"] = df["Churn Category"].fillna("Not Applicable")
df["Churn Reason"] = df["Churn Reason"].fillna("Not Applicable")
```

```python
df.isnull().sum().sum()
```

```python
neg_mask = df["Monthly Charge"] < 0
print("Negative rows:", neg_mask.sum())      # should be 120

df["Monthly Charge Fixed"] = np.where(neg_mask, "Yes", "No")   # flag

estimated = (df["Total Charges"] / df["Tenure in Months"]).round(2)
df.loc[neg_mask, "Monthly Charge"] = estimated[neg_mask]

print("Min monthly charge:", df["Monthly Charge"].min())       # should be positive now    
```

```python
calc_rev = (df["Total Charges"] - df["Total Refunds"]
            + df["Total Extra Data Charges"] + df["Total Long Distance Charges"]).round(2)
print("Mismatch rows:", ((calc_rev - df["Total Revenue"]).abs() > 0.05).sum())
```

```python
df = df.merge(df_zipcode, on="Zip Code", how="left")
print(df.shape)
print(df["Population"].isnull().sum())
```

### Feature Engineering

```python
# Churn Flag
df["Churn Flag"] = (df["Customer Status"] == "Churned").astype(int)

# Tenure Group
df["Tenure Group"] = pd.cut(df["Tenure in Months"],
                            bins=[0, 12, 24, 48, 72],
                            labels=["0-12 Months", "13-24 Months", "25-48 Months", "49-72 Months"],
                            include_lowest=True).astype(str)

# Age Group
df["Age Group"] = pd.cut(df["Age"],
                         bins=[17, 30, 45, 60, 100],
                         labels=["18-30", "31-45", "46-60", "60+"]).astype(str)

# Avg Revenue Per Month
df["Avg Revenue Per Month"] = (df["Total Revenue"] / df["Tenure in Months"]).round(2)

# Check
print(df.shape)
print(df["Churn Flag"].sum())
print(df["Tenure Group"].value_counts())
print(df["Age Group"].value_counts())
print(df["Avg Revenue Per Month"].min(), df["Avg Revenue Per Month"].max())
```

```python
df.columns = (df.columns.str.strip().str.lower()
              .str.replace(r"[^a-z0-9]+", "_", regex=True).str.strip("_"))
print(df.columns.tolist())
```

```python
print("Shape      :", df.shape)
print("Nulls      :", df.isnull().sum().sum())
print("Duplicates :", df.duplicated().sum())
print("Churn rate :", round(df["churn_flag"].mean() * 100, 2))
```

### Connecting to MySQL

```python
from urllib.parse import quote_plus
from sqlalchemy import create_engine, text
from sqlalchemy.types import VARCHAR

DB_USER = "root"
DB_PASSWORD = "your_password"
DB_HOST = "localhost"
DB_PORT = 3306
DB_NAME = "telecom_churn"

# Create the database if it does not exist
server_engine = create_engine(
    f"mysql+pymysql://{DB_USER}:{quote_plus(DB_PASSWORD)}@{DB_HOST}:{DB_PORT}/"
)
with server_engine.begin() as conn:
    conn.execute(text(f"CREATE DATABASE IF NOT EXISTS {DB_NAME}"))

# Connect to that database
engine = create_engine(
    f"mysql+pymysql://{DB_USER}:{quote_plus(DB_PASSWORD)}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)
with engine.connect() as conn:
    print("Connected:", conn.execute(text("SELECT DATABASE()")).scalar())
```

### Exporting Data to MySQL

```python
text_cols = df.select_dtypes(exclude=["number"]).columns
dtype_map = {col: VARCHAR(100) for col in text_cols}

df.to_sql(
    name="customer_churn",
    con=engine,
    if_exists="replace",
    index=False,
    dtype=dtype_map,
    chunksize=1000,
)

with engine.begin() as conn:
    conn.execute(text("ALTER TABLE customer_churn ADD PRIMARY KEY (customer_id)"))

print("Export complete")
```

### Verifying Export

```python
with engine.connect() as conn:
    rows = conn.execute(text("SELECT COUNT(*) FROM customer_churn")).scalar()
    churn = conn.execute(
        text("SELECT ROUND(AVG(churn_flag)*100, 2) FROM customer_churn")
    ).scalar()
print("Rows in MySQL :", rows)
print("Churn rate %  :", churn)
```
